1. Coin Toss Probability
Write a Python program to generate the sample space for three coin tosses and calculate the probability of:
- a) Getting exactly 3 Heads
- b) Getting exactly 2 Heads
- c) Getting at least 1 Head
Hint: Use itertools.product().

```python
import itertools

# Define the coin outcomes
coin_space = ['H', 'T']

# Generate the sample space for three coin tosses
three_coin_tosses = list(itertools.product(coin_space, coin_space, coin_space))
print(f"Sample Space for 3 tosses: {three_coin_tosses}")
print(f"Total outcomes: {len(three_coin_tosses)}\n")

# a) Getting exactly 3 Heads
exactly_3_heads = [outcome for outcome in three_coin_tosses if outcome.count('H') == 3]
P_3H = len(exactly_3_heads) / len(three_coin_tosses)
print(f"Outcomes with exactly 3 Heads: {exactly_3_heads}")
print(f"P(Exactly 3 Heads) = {P_3H:.4f}\n")

# b) Getting exactly 2 Heads
exactly_2_heads = [outcome for outcome in three_coin_tosses if outcome.count('H') == 2]
P_2H = len(exactly_2_heads) / len(three_coin_tosses)
print(f"Outcomes with exactly 2 Heads: {exactly_2_heads}")
print(f"P(Exactly 2 Heads) = {P_2H:.4f}\n")

# c) Getting at least 1 Head
at_least_1_head = [outcome for outcome in three_coin_tosses if outcome.count('H') >= 1]
P_at_least_1H = len(at_least_1_head) / len(three_coin_tosses)
print(f"Outcomes with at least 1 Head: {at_least_1_head}")
print(f"P(At least 1 Head) = {P_at_least_1H:.4f}")

```

2. Dice Probability
Write a Python program to calculate the probability of getting:
- a) An even number
- b) An odd number
- c) A number greater than 4
- d) A number less than or equal to 3
Use the sample space of a single die.

```python
dice_sample_space = list(range(1, 7))
total_outcomes = len(dice_sample_space)

print(f"Sample Space for a single die: {dice_sample_space}")
print(f"Total outcomes: {total_outcomes}\n")

# a) Getting an even number
even_numbers = [x for x in dice_sample_space if x % 2 == 0]
P_even = len(even_numbers) / total_outcomes
print(f"Outcomes with an even number: {even_numbers}")
print(f"P(Even Number) = {P_even:.4f}\n")

# b) Getting an odd number
odd_numbers = [x for x in dice_sample_space if x % 2 != 0]
P_odd = len(odd_numbers) / total_outcomes
print(f"Outcomes with an odd number: {odd_numbers}")
print(f"P(Odd Number) = {P_odd:.4f}\n")

# c) Getting a number greater than 4
greater_than_4 = [x for x in dice_sample_space if x > 4]
P_gt_4 = len(greater_than_4) / total_outcomes
print(f"Outcomes with a number greater than 4: {greater_than_4}")
print(f"P(Number > 4) = {P_gt_4:.4f}\n")

# d) Getting a number less than or equal to 3
less_than_eq_3 = [x for x in dice_sample_space if x <= 3]
P_lte_3 = len(less_than_eq_3) / total_outcomes
print(f"Outcomes with a number less than or equal to 3: {less_than_eq_3}")
print(f"P(Number <= 3) = {P_lte_3:.4f}")
```

3. Two-Dice Experiment
Generate the sample space for rolling two dice and calculate the probability that:
- a) The sum is 7
- b) The sum is greater than 8
- c) Both dice show the same number
- d) At least one die shows 6
Hint: Use itertools.product(range(1,7), range(1,7)).

```python
import itertools

# Generate the sample space for rolling two dice
dice_outcomes = range(1, 7)
two_dice_sample_space = list(itertools.product(dice_outcomes, dice_outcomes))
total_outcomes_two_dice = len(two_dice_sample_space)

print(f"Sample Space for two dice: {two_dice_sample_space}")
print(f"Total outcomes: {total_outcomes_two_dice}\n")

# a) The sum is 7
sum_is_7 = [outcome for outcome in two_dice_sample_space if sum(outcome) == 7]
P_sum_7 = len(sum_is_7) / total_outcomes_two_dice
print(f"Outcomes where sum is 7: {sum_is_7}")
print(f"P(Sum = 7) = {P_sum_7:.4f}\n")

# b) The sum is greater than 8
sum_gt_8 = [outcome for outcome in two_dice_sample_space if sum(outcome) > 8]
P_sum_gt_8 = len(sum_gt_8) / total_outcomes_two_dice
print(f"Outcomes where sum is greater than 8: {sum_gt_8}")
print(f"P(Sum > 8) = {P_sum_gt_8:.4f}\n")

# c) Both dice show the same number
same_number = [outcome for outcome in two_dice_sample_space if outcome[0] == outcome[1]]
P_same_number = len(same_number) / total_outcomes_two_dice
print(f"Outcomes where both dice show the same number: {same_number}")
print(f"P(Same Number) = {P_same_number:.4f}\n")

# d) At least one die shows 6
at_least_one_6 = [outcome for outcome in two_dice_sample_space if 6 in outcome]
P_at_least_one_6 = len(at_least_one_6) / total_outcomes_two_dice
print(f"Outcomes where at least one die shows 6: {at_least_one_6}")
print(f"P(At least one 6) = {P_at_least_one_6:.4f}")
```

5. Law of Large Numbers
Simulate 10,000 fair coin tosses and calculate the cumulative probability of Heads after every toss.
Plot:
- X-axis → Number of tosses
- Y-axis → Cumulative probability of Heads
- Horizontal line → \(P(H)=0.5\)
Use a logarithmic X-axis.

```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Simulate 10,000 fair coin tosses (0 = Tails, 1 = Heads)
n_tosses = 10000
tosses = np.random.binomial(1, 0.5, n_tosses)

# Calculate cumulative probability of Heads after every toss
cumulative_heads = np.cumsum(tosses)
trials = np.arange(1, n_tosses + 1)
cumulative_prob = cumulative_heads / trials

# Plotting the results
plt.figure(figsize=(10, 5))
plt.plot(trials, cumulative_prob, color='steelblue', label='Empirical Probability')
plt.axhline(0.5, color='red', linestyle='--', linewidth=1.5, label='True P(H) = 0.5')

# Configure axes and labels
plt.xscale('log')
plt.xlabel('Number of tosses (log scale)')
plt.ylabel('Cumulative probability of Heads')
plt.title('Law of Large Numbers: Cumulative Probability of Heads')
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

6. Conditional Probability with Dice
Roll a die.
Define:
- \(A\): number is even
- \(B\): number is greater than 3
- Write a program to calculate: P(A|B)
- Verify the result manually.

```python
# Define the sample space for a single die
dice_sample_space = list(range(1, 7))

# Define event A: number is even
A = [x for x in dice_sample_space if x % 2 == 0]
print(f"Event A (even numbers): {A}")

# Define event B: number is greater than 3
B = [x for x in dice_sample_space if x > 3]
print(f"Event B (numbers greater than 3): {B}")

# Calculate P(A and B) - the intersection of A and B
# This means numbers that are both even AND greater than 3
intersection_A_B = [x for x in A if x in B]
print(f"Intersection of A and B (A ∩ B): {intersection_A_B}")

P_A_and_B = len(intersection_A_B) / len(dice_sample_space)
print(f"P(A ∩ B) = {P_A_and_B:.4f}")

# Calculate P(B)
P_B = len(B) / len(dice_sample_space)
print(f"P(B) = {P_B:.4f}")

# Calculate P(A|B) using the formula P(A|B) = P(A and B) / P(B)
if P_B != 0:
    P_A_given_B = P_A_and_B / P_B
    print(f"P(A|B) = P(A ∩ B) / P(B) = {P_A_given_B:.4f}")
else:
    print("Cannot calculate P(A|B) as P(B) is zero.")

# Manual Verification:
# Sample space = {1, 2, 3, 4, 5, 6}
# Event A (even) = {2, 4, 6}
# Event B (greater than 3) = {4, 5, 6}
# A and B = {4, 6}
# Given B has occurred, the new sample space is {4, 5, 6}. Out of these, the numbers that are even are {4, 6}.
# So, P(A|B) = 2 (favorable outcomes) / 3 (total outcomes in B) = 2/3 ≈ 0.6667
```

7. Bayes' Theorem - Medical Test Problem
A disease affects 2% of a population.
A diagnostic test has:
- Sensitivity = 95%
- False-positive rate = 4%
- Write a Python program to calculate: P(Disease|Positive)
- Display the answer as a percentage.

```python
# Define the given probabilities
P_Disease = 0.02  # P(D) - Disease affects 2% of a population
Sensitivity = 0.95  # P(Pos|D) - Sensitivity: P(Positive | Disease)
False_Positive_Rate = 0.04  # P(Pos|ND) - False-positive rate: P(Positive | No Disease)

# Calculate P(No Disease)
P_No_Disease = 1 - P_Disease

# Calculate P(Positive) using the law of total probability
# P(Positive) = P(Positive|Disease) * P(Disease) + P(Positive|No Disease) * P(No Disease)
P_Positive = (Sensitivity * P_Disease) + (False_Positive_Rate * P_No_Disease)

# Calculate P(Disease|Positive) using Bayes' Theorem
# P(D|Pos) = [P(Pos|D) * P(D)] / P(Pos)
if P_Positive != 0:
    P_Disease_given_Positive = (Sensitivity * P_Disease) / P_Positive
    print(f"P(Disease | Positive) = {P_Disease_given_Positive:.4f} ({P_Disease_given_Positive * 100:.2f}%)")
else:
    print("Cannot calculate P(Disease|Positive) as P(Positive) is zero.")
```

8. Email Classification
- Suppose: 20% of emails are spam.
- A spam email contains the word "offer" with probability 80%.
- A non-spam email contains offer with probability 10%.
- Using Bayes theorem, calculate: P(Spam|offer)
- Write a Python program to display the result.

```python
# Define the given probabilities
P_Spam = 0.20  # P(S) - 20% of emails are spam
P_Offer_given_Spam = 0.80  # P(O|S) - Spam email contains "offer" with probability 80%
P_Offer_given_NotSpam = 0.10  # P(O|NS) - Non-spam email contains "offer" with probability 10%

# Calculate P(Not Spam)
P_NotSpam = 1 - P_Spam

# Calculate P(Offer) using the law of total probability
# P(Offer) = P(Offer|Spam) * P(Spam) + P(Offer|Not Spam) * P(Not Spam)
P_Offer = (P_Offer_given_Spam * P_Spam) + (P_Offer_given_NotSpam * P_NotSpam)

# Calculate P(Spam|Offer) using Bayes' Theorem
# P(S|O) = [P(Offer|Spam) * P(Spam)] / P(Offer)
if P_Offer != 0:
    P_Spam_given_Offer = (P_Offer_given_Spam * P_Spam) / P_Offer
    print(f"P(Spam | Offer) = {P_Spam_given_Offer:.4f} ({P_Spam_given_Offer * 100:.2f}%)")
else:
    print("Cannot calculate P(Spam|Offer) as P(Offer) is zero.")
```
